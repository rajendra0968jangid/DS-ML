```python
import numpy as np
scores = np.array([88,34,76,85,99,80])
# mean , min , max
print("average",scores.mean())
print("min",scores.min())
print("max",scores.max())
print(scores > 60)
print(scores + 2)
print(scores * 2)
print(scores.shape) # ndim
print(scores.dtype)
print(scores.size)
```
