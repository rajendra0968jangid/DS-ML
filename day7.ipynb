```python
# 4 students and 3 subjects
import numpy as np
arr1 = np.array([
    [70,80,90],
    [60,75,85],
    [88,92,95],
    [58,42,65]
])
print(arr1)
print(arr1[0,1]) # row 0, cols 1
```

```python
# avg marks of each students
avg_marks_students = np.mean(arr1,axis=1) # axis=1 for each row
print("avg marks of each students",np.round(avg_marks_students))

# avg marks of each subjects
avg_marks_subjects = np.mean(arr1,axis=0) # axis=1 for each row
print("avg marks of each subjects",np.round(avg_marks_subjects))
```

```python
# total marks of each students  # (students are in row format that'why we are using axis=1)
student_total = np.sum(arr1,axis=1)
print("total marks of each students",student_total)

# highest mark in each subject # (subjects are in cols format that'why we are using axis=0)
highest_sub_marks = np.max(arr1,axis=0)
print("highest mark in each subject",highest_sub_marks)
```

```python
# sorting
arr = np.array([10,3,15,20])
print(arr)
print(np.sort(arr)) # ascending order
print(np.sort(arr)[::-1]) # decending order
```

```python
# sorting 2d array
import numpy as np
arr2 = np.array([
    [3,2,1],
    [6,5,7]
])
print(np.sort(arr2)) # by default row wise
print(np.sort(arr2,axis=0)) # column wise sort
```

```python
# filtering in numpy
arr = np.array([10,20,25,15,40])
print(arr[arr > 20])
print(arr[arr%2 == 0]) # even numbers
```

```python
# Dynamic array means: values come at runtime (from user, loop, file, etc.).
import numpy as np
n = int(input("Enter size: "))
arr = []
for i in range(n):
  val = int(input("Enter value: "))
  arr.append(val)

np_arr = np.array(arr)
print(np_arr)
print(type(np_arr))
```
