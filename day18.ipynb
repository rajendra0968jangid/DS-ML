```python
# import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load data
url = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/movies_tv_shows_raw.csv"
df = pd.read_csv(url)
```

```python
print(df)
```

```python
# basic data understand steps
# print starting 5 entrys
print(df.head())
```

```python
# print shape
print(df.shape)
```

```python
# columns types
print(df.dtypes)
```

```python
# print ending 5 entrys
print(df.tail())
```

```python
# missing values
print(df.isna().sum())
```

```python
# duplicates rows
print(df.duplicated(subset='show_id').sum())
```

```python
# clean text fields
text_cols = ['type','title','director','cast','country','rating','duration','listed_in']
for col in text_cols:
  df[col] = df[col].astype('string').str.strip()
```

```python
print(df.dtypes)
```

```python
print(df['director'])
```

```python
# convert empty or missing values -> na
missing_tokens = ["","N/A","NA","None","Unknown","nan"]
for col in text_cols:
  df[col] = df[col].replace(missing_tokens,pd.NA)
```

```python
print(df.tail())
```

```python
# standardize category
# df['type'] = df['type'].str.title()
print(df['type'].unique())
```

```python
df['rating'] = df['rating'].str.upper()
print(df['rating'].unique())
```

```python
print(df['country'].unique())
```

```python
# fix release year
# df.loc[27] # df['release_year'] -> 2019
df.loc[27,'release_year'] = 2019
```

```python
# release_year and date_added
df['date_added'] = pd.to_datetime(df['date_added'] , errors='coerce')
df['release_year'] = pd.to_numeric(df['release_year'] , errors='coerce')
```

```python
print(df.dtypes)
```

```python
df.loc[6,'director'] = 'Sergey A.'
df.loc[30,'director'] = 'Sanaa Akroud'
df.loc[30,'cast'] = 'Sanaa Akroud'
df.loc[30,'country'] = 'Canada'
```
