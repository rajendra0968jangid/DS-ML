Filter operations

```python
import pandas as pd

data = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/student-data.json"

df = pd.read_json(data)

print(df)
```

```python
filter_data = df["maths"] > 50

# print(filter_data)

# print(df[filter_data])

print(df[df["maths"] > 50])
```

```python
# english < 60

print(df[df["english"] < 60])
```

```python
# physics marks equal to  78
print(df[df["physics"] == 78])
```

```python
# physics marks not equal to  78
print(df[df["physics"] != 78])
```

```python
# name equal to Sundar
print(df[df["name"] == "Sundar"])
```

```python
# print users where maths marks > 60 and english marks < 60
print(df[ (df["maths"] > 60) & (df["english"] < 60)])
```

```python
# print users where gender is male and maths marks > 60
print(df[(df["gender"] == "Male") & (df["maths"] > 60)])
```

```python
# print users where gender is male or maths marks > 60
print(df[(df["gender"] == "Male") | (df["maths"] > 60)])
```

```python
print(df.query("maths > 60 or english < 60"))
```

```python
# physics > 60 and english < 60 then =>>> physics
print(df[(df["physics"] > 60) & (df["english"] < 60)]["physics"])
```

```python
print(df[(df["physics"] > 60) & (df["english"] < 60)][["physics","maths"]])
```

handle missing values

```python
import pandas as pd
url = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/file2.json"

df = pd.read_json(url)

print(df)
```

```python
# isnull()
print(df.isnull())
```

```python
# isnull() with sum()
print(df.isnull().sum())
```

```python
# isnull() with sum() -> row wise
print(df.isnull().sum(axis=1))
```

```python
# isnull() with sum() -> cols wise
print(df.isnull().sum(axis=0))
```

```python
# drop values by row
print(df.dropna())
```

```python
# drop null values cols
print(df.dropna())
```

```python
# fill
print(df.fillna(0))
```

```python
# fill with 100
print(df.fillna(100))
```

```python
# fill rollno -> 300 | marks -> 400
df["roll no"] = df["roll no"].fillna(300)
print(df)
```

```python
# fill 400 in marks
df["marks"] = df["marks"].fillna(400)
print(df)
```

```python
# alternative option |
df = pd.read_json(url)
df.fillna({"roll no": 200, "marks":500},inplace=True)
print(df)
```

```python
# marks cols -> nan -> mean fillna
df = pd.read_json(url)
mm = df["marks"].mean()
print(mm)
df["marks"] = df["marks"].fillna(mm)
print(df)
```
