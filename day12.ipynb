```python
import pandas as pd
data = {
    "Emp_ID": [201, 202, 203, 204, 205, 206],
    "Name": ["Arjun", "Priya", "Karan", "Meena", "David", "Anjali"],
    "Department": ["Marketing", "IT", "HR", "Finance", "Sales", "IT"],
    "Salary": [58000, 62000, 47000, 65000, 51000, 59000],
    "Experience": [3, 5, 2, 6, 2, 4]
}
df = pd.DataFrame(data)

print(df)
```
