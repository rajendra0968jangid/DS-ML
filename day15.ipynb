```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'student': [
        'Rahul', 'Priya', 'Amit', 'Neha', 'Rohit',
        'Sneha', 'Karan', 'Pooja', 'Vikas', 'Anjali'
    ],
    'gender': [
        'Male', 'Female', 'Male', 'Female', 'Male',
        'Female', 'Male', 'Female', 'Male', 'Female'
    ],
    'study_hours': [2, 5, 3, 6, 1, 4, 2, 5, 3, 4],
    'attendance': [70, 90, 75, 95, 60, 85, 65, 92, 80, 88],
    'math': [55, 85, 65, 92, 40, 78, 50, 88, 70, 82],
    'science': [60, 88, 70, 90, 45, 80, 55, 85, 72, 86],
    'english': [65, 92, 68, 94, 50, 82, 48, 90, 75, 84]
}
df = pd.DataFrame(data)
print(df)
```

```python
# new column
df['total_marks'] = df['math'] + df['science'] + df['english']
print(df)
```

```python
# average (mean) marks of each student
df['avg_marks'] = df[['math','science','english']].mean(axis=1)
print(df)
```

```python
# top 3 students  (good score) -> total_marks
print(df.sort_values('total_marks').tail(3)) # ascending -> sorting
```

```python
# bottom 3 students (bad score) -> total_marks
print(df.sort_values('total_marks').head(3))
```

```python
# top 3 students  (good score) -> total_marks
print(df.sort_values('total_marks',ascending=False).head(3)) # ascending -> sorting
```

```python
# bottom 3 students (bad score) -> total_marks
print(df.sort_values('total_marks').head(3))
```

```python
# find student who scored above 75 on average:-
# print(df['avg_marks'] > 75) # filter out
print(df[df['avg_marks'] > 75]) # print filter out data
```

```python
# calculate avg marks by gender
# male -> avg -> avg
# female -> avg -> avg

# common avg marks
print(df['avg_marks'].mean()) # 72 avg | common of both gender

# group by
print(df.groupby('gender')['avg_marks'].mean()) # gender wise

# common max marks
print(df['total_marks'].max())  # common of both gender

# gender wise max
print(df.groupby('gender')['total_marks'].max())
```

```python
# missing values
print(df.isnull())
```

```python
# missing values sum
print(df.isnull().sum())
```

```python
# check for duplicates values:-
print(df.duplicated())
```

```python
# check for duplicates values:-
print(df.duplicated().sum())
```

```python
# numpy for numerical calcal:
marks = df['avg_marks']
print("Average marks:", np.mean(marks))
print("Highest marks:", np.max(marks))
print("Lowest marks:", np.min(marks))
print("Standard deviation:", np.std(marks))
```

```python
#Count students scoring at least 75 on average:
count = np.sum(marks >= 75)

print("Students scoring 75 or above:", count)
```

```python
#Visualize the results
plt.figure(figsize=(10, 5))

plt.bar(df['student'], df['avg_marks'])

plt.title('Average Marks of Students')
plt.xlabel('Student')
plt.ylabel('Average Marks')
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
