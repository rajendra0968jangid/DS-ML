```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
tips = tips.rename(columns={"sex":"gender"})
print(tips)
```

Matrix plot

```python
#heatmap
# clustermap
```

```python
import plotly.express as px # demi data sets
gap = px.data.gapminder()
print(gap.head())
```

```python
tmp_data = gap[gap['country']=='India']
print(tmp_data)
```

```python
# line plot
tmp_df = gap[gap['country']=='India'] # pandas filter
# print(tmp_df)
sns.lineplot(data=tmp_df,x='year',y='lifeExp')
```

```python
# heatmap
tmp_df = gap.pivot(index='country',columns='year',values='lifeExp')
print(tmp_df)
```

```python
sns.heatmap(data=tmp_df)
```

```python
# filter data -> i want only asia -> country
tmp_df = gap[gap['continent'] == 'Asia']
# print(tmp_df)
tmp_df_updated = tmp_df.pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(data=tmp_df_updated,annot=True,linewidth=0.1,cmap='summer')
```

Categorical plots

```python
# figure level
# 1. catplot

## these all are axes level

# categorical scatter plot ->
# 1. stripplot
# 2. swarmplot
# categorical distribution plot ->
# 1. boxplot
# 2. violinplot
```

```python
import seaborn as sns
tips = sns.load_dataset('tips')
tips = tips.rename(columns={"sex":"gender"})
print(tips)
```

```python
# strip plot
sns.stripplot(data=tips,x='day',y='total_bill')
```

```python
# scatter plot
sns.scatterplot(data=tips,x='tip',y='total_bill')
```

```python
# figure level
sns.catplot(data=tips,x='day',y='total_bill',kind='strip')
```

```python
#hue
sns.catplot(data=tips,x='day',y='total_bill',kind='strip',hue='gender')
```
