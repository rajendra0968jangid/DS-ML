```python
# relation
# bivariate -> works on 2 var
# scatterplot
# lineplot
# relplot -> by the help of relplot you creats scatterplot and lineplot
```

```python
tips = sns.load_dataset('tips')
print(tips)
```

```python
print(tips.head())
```

```python
# scatter plot -> axes level function
tips = tips.rename(columns={"sex":"gender"})
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='gender')
```

```python
# relplot
sns.relplot(data=tips,x='total_bill',y='tip',hue='gender',kind='scatter')
```

```python
# lineplot
sns.lineplot(data=tips,x='total_bill',y='tip',hue='gender')
```

```python
# lineplot using relplot
sns.relplot(data=tips,x='total_bill',y='tip',hue='gender',kind='line')
```

```python
# scatterplot using relplot
sns.relplot(data=tips,x='total_bill',y='tip',hue='gender',kind='scatter',style='time')
```

```python
# scatterplot using relplot
sns.relplot(data=tips,x='total_bill',y='tip',col='gender',kind='scatter',style='time')
```

```python
# scatterplot using relplot
sns.relplot(data=tips,x='total_bill',y='tip',row='gender',kind='scatter',style='time')
```

```python
# scatterplot using relplot
sns.relplot(data=tips,x='total_bill',y='tip',col='gender',kind='scatter',row='time')
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',col='day',kind='scatter',col_wrap=3)
```
