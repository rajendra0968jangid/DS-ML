```python
import numpy as np
arr = np.arange(12)
print(arr)

# reshape 2d
arr1 = np.arange(12) # 3*4 | 4*3 | 6*2 | 2*6  (rows,cols)
res_arr1 = np.reshape(arr1,(3,4))
print(res_arr1)

# reshape 3d
arr2 = np.arange(24) # 3*2*4 | 4*2*3 |
res_arr2 = np.reshape(arr2,(3,2,4))
print(res_arr2)
```

```python
# flatten | create copy then works
import numpy as np
arr = np.arange(12).reshape(3,4)
print(arr)

# 2d -> 1d
flat_arr = arr.flatten()
flat_arr[0] = 100
print(flat_arr)
print(arr)

# 3d -> 1d
arr1 = np.arange(24).reshape(3,4,2)
print(arr1)
flat_arr1 = arr1.flatten()
print(flat_arr1)
```

```python
# ravel | work on original array
import numpy as np
arr = np.arange(12).reshape(3,4)
print(arr)

# 2d -> 1d
flat_arr = arr.ravel()
flat_arr[0] = 100
print(flat_arr)
print("updated array:",arr)

# 3d -> 1d
```

```python
# transpose
import numpy as np
arr = np.arange(12).reshape(3,4)
print(arr)
trans_arr = arr.T # .transpose()
print(trans_arr) # 4 X 3
```

```python
# adding and removing data
# 1. np.concatenate((arr1,arr2))
# 2. np.vstack((original,new_arr)) # add new row
# 3. np.hstack((original,new_arr)) # add new cols
# 4. np.delete(arr,index_no) # delete element by index number
```

```python
# np.concatenate
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])

# np.concatenate
result = np.concatenate((arr1,arr2))
print(result)

#
result1 = arr1 + arr2
print(result1)
```

```python
# 2d array
import numpy as np
arr1 = np.array([
    [10,20],
    [30,40]
])
arr2 = np.array([
    [50,60],
    [70,80]
])
result = np.concatenate((arr1,arr2)) # row - wise
print(result)

result = np.concatenate((arr1,arr2),axis=1) # col - wise
print(result)
```

```python
# add new row for 1d
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])

# vstack
result = np.vstack((arr1,arr2))
print(result)
```

```python
# add new row for 2d
original = np.array([
    [10,20,30],
    [40,50,60]
])
new_row = np.array([70,80,90])

# alternative of vstack
result = np.concatenate((original,[new_row])) # axis=0
print(result)

# vstack
result = np.vstack((original,new_row))
print(result)
```

```python
# hstack -> add new cols
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])

result = np.hstack((arr1,arr2))
print(result)
```

```python
# handle missing values
# np.nan | np.isnan
import numpy as np
arr = np.array([10,20,np.nan,40,np.nan,60]) # null -> nan
print(arr)

#
result = np.isnan(arr)
print(result)
```

```python
import numpy as np
arr = np.array([10,20,np.nan,40,np.nan,60]) # null -> nan
print(arr)

#
index = np.where(np.isnan(arr)) # for index
print(index)

#
value = arr[index] # for value
print(value)
```

```python
# count number of nan values
import numpy as np
arr = np.array([10,20,np.nan,40,np.nan,60]) # null -> nan
print(arr)

result = np.sum(np.isnan(arr))
print(result)
```

```python
# np.any -> to check any nan value is exist or not
import numpy as np
arr = np.array([10,20,np.nan,40,np.nan,60]) # null -> nan
print(arr)

result = np.any(np.isnan(arr))
print(result)
```

```python
# np.all -> to check all value for nan or not
import numpy as np
arr = np.array([10,20,np.nan,40,np.nan,60]) # null -> nan
print(arr)

result = np.all(np.isnan(arr))
print(result)
```
