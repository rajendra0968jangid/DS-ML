```python
# Problem: How do you initialize a 3×4 integer matrix populated entirely with zeros?
import numpy as np
zero_mat = np.zeros((3,4),dtype=int)
print(zero_mat)
print("shape",zero_mat.shape)
print("dtype",zero_mat.dtype)
```

```python
# Problem: How do you create a 3D tensor of shape (2, 3, 2) filled with ones?
import numpy as np
ones_tensor = np.ones((2,3,2),dtype=np.float32)
print(ones_tensor)
print("tensor shape",ones_tensor.shape)
print("data type", ones_tensor.dtype)
```

```python
# Problem: How do you construct a 3×3 matrix where every element is set to 7.5 ?
import numpy as np
cont_mat = np.full(shape=(3,3),fill_value=7.5)
print(cont_mat)
```

```python
# Problem: How does the arr.fill() method differ from np.full(), and how is it used to mutate an existing pre-allocated array?
import numpy as np
buf = np.empty((2,3),dtype=np.int32)
print(buf)

# Overwrite all values in-place without new memory allocation
buf.fill(99)
print(buf)
```

```python
# Problem: How do you generate an array of numbers from 10 to 50 with a step of 5,
# and an array with floating-point increments from 0.0 to 1.0 with a step of 0.2?

import numpy as np
int_seq = np.arange(10,50,5) # start,stop,step
print(int_seq)

float_seq = np.arange(0.0 , 1.0 , 0.2) #
print(float_seq)
```

```python
#Array Properties (shape, size, ndim, dtype) # 1D or 2D and 3D
import numpy as np
vec = np.array([1,2,3])
mat = np.array([[1,2,3],[4,5,6]])
ten = np.array([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
])
print("shape",vec.shape)
print(mat.shape)
print(ten.shape)

print("size",vec.size)
print(mat.size)
print(ten.size)

print("ndim",vec.ndim)
print(mat.ndim)
print(ten.ndim)
```

```python
# indexing and slicing
# 1d array
# indexing is always start from zero and position is always start from one
import numpy as np
marks = np.array([70,80,90,85])
print(marks)
print(marks[0]) # 70
print(marks[2]) # 90
print(marks[-1]) # negative one for last value
```

```python
# 2d array indexing
import numpy as np
students = np.array([
    [70,80,90],
    [65,75,85]
])
print(students)
# print(students[rows][cols])
print("row 0 = ",students[0])
print("row 0 and cols 0 = ",students[0][0])
```

```python
# question
import numpy as np
n =  int(input("Enter number of students:"))
names = []
marks = []

for i in range(n):
  name = input("Enter Student name:")
  mark = int(input("Enter marks:"))

  names.append(name)
  marks.append(mark)

# convert list in numpy array
names_array = np.array(names)
marks_array = np.array(marks)
print("Names:", names_array)
print("Marks:", marks_array)
```

```python
# slicing
# 1d array
import numpy as np
arr = np.array([10,20,30,40,50,60])
print(arr)
print("slicing",arr[1:3]) # starting index : end index (last value exclude)
```

```python
# basic slicing concepts
# note:- if you update slicing array then your original array also will changed
arr1 = np.array([5,6,7,9,11,2])
print(arr1)
arr1_slice = arr1[1:4]
print(arr1_slice)
print("before update index 0 of slice = ",arr1_slice[0])
arr1_slice[0] =  20
print("after update index 0 of slice = ",arr1_slice[0])

print("original array =", arr1)
```

```python
# 1d example
import numpy as np
arr1 = np.array([5,6,7,8,2,3])
arr1_slice = arr1[:4] # default starting from zero index
print(arr1)
print(arr1_slice)
```

```python
# 1d example
import numpy as np
arr1 = np.array([5,6,7,8,2,3])
arr1_slice = arr1[2:] # default last index number
print(arr1)
print(arr1_slice)
```

```python
# 1d example
import numpy as np
arr1 = np.array([5,6,7,8,2,3])
arr1_slice = arr1[:] # default start from 0 and last index
print(arr1)
print(arr1_slice)
```

```python
# 1d example
import numpy as np
arr1 = np.array([5,6,7,8,2,3])
arr1_slice = arr1[1:4:2] #  start from 1 and end with 3 and step by 1
print(arr1)
print(arr1_slice)
```

```python
# 1d example
import numpy as np
arr1 = np.array([5,6,7,8,2,3])
arr1_slice = arr1[::2] #  start from 0 and end with last and step by 1
print(arr1)
print(arr1_slice)
```

```python
# 1d example reverse
import numpy as np
arr1 = np.array([5,6,7,8,2,3])
arr1_slice = arr1[::-1] #  start from 0 and end with last and step by -1
print(arr1)
print(arr1_slice)
```

```python
# 1d example basic
import numpy as np
arr1 = np.array([5,6,7,8,2,3])
print(arr1)
arr1_slice = arr1[1:4]
print("before update=",arr1_slice)

# update slice array
arr1_slice[0] = 21
print("after update=",arr1_slice)

print("original array",arr1) #
```

```python
# fancy indexing for 1d
# note:- its create a new array instead of changes in original one
import numpy as np
arr1 = np.array([5,6,7,8,2,3])
print(arr1)
arr1_slice = arr1[[1,2,3]] # fancy indexing
print("before update=",arr1_slice)

# update slice array
arr1_slice[0] = 21
print("after update=",arr1_slice)

print("original array",arr1) #
```
