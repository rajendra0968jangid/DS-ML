```python
import statistics

data = [4, 2, 7, 4, 9, 4, 5]

# Calculate statistics
mean_val = statistics.mean(data)
median_val = statistics.median(data)
mode_val = statistics.mode(data)

print("Mean:", mean_val)    # Output: 5.0
print("Median:", median_val)  # Output: 4
print("Mode:", mode_val)    # Output: 4
```

```python
x = int(input("Enter length of items: "))
y = []
for i in range(x):
  z = int(input(f"number {i} = "))
  y.append(z)
print(x)
print(y)
y.sort()
print(y)

# if number of elements is even
if (x%2 == 0): # even -> true
  print("if part run",(x+1)//2)
  m = x//2
  right = y[m]
  print(right)
  left = y[m-1]
  print(left)
  median = (right+left)/2
  print(median)
else:
  print(y[(x//2)])
```
