```python
# eco sales anal.

import numpy as np

# note:- because no up and down 0 | 1000
np.set_printoptions(suppress=True)

sales = np.array([
    [1001, 101, 2, 500, 10, 4.5, 3], # order_id,product_id,quantity,price,discount,rating,delivery_day
    [1002, 102, 1, 1200, 5, 4.2, 5],
    [1003, 103, 4, 250, 0, 4.8, 2],
    [1004, 104, 3, 800, 15, 3.9, 7],
    [1005, 105, 5, 150, 10, 4.1, 4],
    [1006, 101, 2, 500, 20, 4.6, 3],
    [1007, 106, 1, 2500, 5, 4.9, 2],
    [1008, 103, 3, 250, 0, 3.8, 8],
    [1009, 107, 2, 1800, 10, 4.7, 4],
    [1010, 108, 6, 300, 15, 3.5, 6],
    [1011, 105, 3, 150, 5, 4.3, 3],
    [1012, 109, 1, 3500, 20, 4.9, 2],
    [1013, 110, 4, 450, 10, 4.0, 5],
    [1014, 106, 2, 2500, 15, 4.8, 3],
    [1015, 108, 5, 300, 0, 3.2, 9]
])
print(sales)
```

```python
# shape,size , dimensions, number of rows|orders , number of columns
print(sales.shape)
print(sales.size)
print(sales.ndim)

t = sales.shape
print("rows",t[0])
print("columns",t[1])

print(len(sales))
print(len(sales[0]))
```

```python
# extract quantity -> reshape(-1,1)
quantity = sales[:,2]
print(quantity)
```

```python
# First 5 orders
print(sales[:5]) # print(sales[0:5])
```

```python
# Last 3 orders
print(sales[-3:])
```

```python
# calculate gross amount (quantity * price)
quantity = sales[:,2]
price = sales[:,3]

gross = quantity * price

print(gross)
```

```python
# calculate discount amount -> (gross * discount)/100
discount = sales[:,4]

quantity = sales[:,2]
price = sales[:,3]

gross = quantity * price

discount_amt = gross * discount / 100

print(discount_amt)
```

```python
# calculate final revenue -> (gross - discount_amt)
rev = gross - discount_amt

print(rev)
```

```python
# total revenue -> np.sum
print(np.sum(rev))
```

```python
# average order revenue -> np.mean
print(np.mean(rev))
```

```python
# highest revenue
print(np.max(rev))
```

```python
# lowest revenue
print(np.min(rev))
```

```python
# total quantity sold
print(np.sum(quantity))
```

```python
# find higly rated orders -> >= 4.5 --> fancy indexing | boolean indexing
rating = sales[:,5]

result = sales[rating >= 4.5]

print(result)
print(np.mean(rating))
```

```python
# discount_amt -> gross * discount/100
# gross -> ( quantity * price )
# revenue -> (gross - disount_amt)

# revenue -> sort -> asc | desc

gross = sales[:,6] * sales[:,3]

discount_amt = gross * sales[:,4]/100

rev = gross - discount_amt
print(rev)
print(np.sort(rev))
print(np.sort(rev)[::-1])
```

```python
# add new order | row -> vstack()
new_order = np.array([1016,111,2,600,10,4.6,5])

sales = np.vstack((sales,new_order))
print(sales)
```

```python
# # # # add new column -> gross -> hstack()
gross = sales[:,2] * sales[:,3]

sales = np.hstack((sales,gross.reshape(-1,1)))
print(sales)
```

```python
# # add new column -> revenue -> hstack()
gross = sales[:,2] * sales[:,3]
discount_amt = gross * sales[:,4]/100
rev = gross - discount_amt
sales = np.hstack((sales,rev.reshape(-1,1)))

print(sales)
```

```python
# delete rating column
sales = np.delete(sales,5,axis=1)
print(sales)
```

```python
# delete order -> 1005
result = np.delete(sales,4,axis=0)

print(result)
```

```python
# [2,5]-> nan | [7,5] -> nan
sales_missing = sales.copy()

sales_missing[2,5] = np.nan
sales_missing[7,5] = np.nan

print(sales_missing)
```
