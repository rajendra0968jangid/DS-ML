```python
# keep a copy of original df
raw_df = df.copy()
```

```python
# Transform
# student_id -> string | student_name -> string | gender -> string
# age -> int | study_hours_per_day  -> float
# assessment_date  -> date

# method 1
df['student_id'] = df['student_id'].astype('string')
df['student_name'] = df['student_name'].astype('string')
df['gender'] = df['gender'].astype('string')

# method 2
cols = ['student_id','student_name','gender']
for col in cols:
  df[col] = df[col].astype('string')
```

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
```

```python
df['assessment_date'] = pd.to_datetime(df['assessment_date'], errors='coerce')
```
