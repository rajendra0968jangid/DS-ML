```python
# enumerate()
names = ["hello","world","raj","amit"]
for index,name in enumerate(names):
  print(index,name)
```

```python
names = ["raj","hello"]
marks = [80,90]
for name,mark in zip(names,marks):
  print(name,mark)

for i in range(len(names)):
  print(names[i],marks[i])
```

```python
# function
def greet():
  print("helo")

greet()
```

```python
# function with parameter
def greet(name):
  print("hello",name)

greet("raj")
```

```python
# function with return
def add(a,b):
  return a + b

result = add(10,20)
print(result)
```

```python
# function with default parameter
def greet(name="Guest"):
  print("hello",name)

greet()
greet("raj")
```

```python
# keyword arguments
def student(name,age):
  print(name,age)

student("hello",20)

student(30,"amit")
#alternative
student(age=30,name="amit")
```

```python
# *args
def add_numbers(*numbers): # single paramter
  print(numbers[0])
  print(type(numbers))

  for i in numbers[0]:
    print(i)


# add_numbers(10,20,30) # multiple arguments
add_numbers([{"name":"hello"},{"name":"amit"}]) # hello and amit
```

```python
# list slicing
numbers = [10,20,30,40,50]
print(numbers[1]) # 20
print(numbers[4]) # 50
print(numbers[1:4]) # 1 , 2, 3
print(numbers[::-1])
```
