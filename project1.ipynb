```python
import numpy as np
patients = np.array([
    [101, 25, 120, 80, 98.6, 72], # patient Id , age , systolic BP , diastolic bp , temp , heart rate
    [102, 34, 135, 88, 99.1, 85],
    [103, 29, np.nan, 76, 98.4, 70],
    [104, 45, 150, np.nan, 101.2, 92],
    [105, 52, 160, 95, 99.8, np.nan],
    [106, 31, 128, 82, 98.7, 76],
    [107, 40, np.nan, 90, 100.5, 88],
    [108, 27, 118, 78, 98.2, 68],
    [109, 56, 170, 100, np.nan, 95],
    [110, 38, 140, 85, 99.3, 80]
])
print(patients)
```

```python
# find missing values -> np.isnan()
print(np.isnan(patients))
```

```python
# sum of nan values
print(np.sum(np.isnan(patients))) # nan  | missing
```

```python
# find the shape of the dataset
print(patients.shape) # rows -> 10 | cols-> 6
```

```python
# find missing values column wise
missing = np.isnan(patients).sum(axis=0)
# row -> 0 | cols -> 1

print(missing) # patient Id , age , systolic BP , diastolic bp , temp , heart rate
```

```python
# find average systolic bp ingoring nan -> np.nanmean()
max_sys_bp = np.nanmax(patients[:,2])
print(max_sys_bp)

min_sys_bp = np.nanmin(patients[:,2])
print(min_sys_bp)

avg_sys_bp = np.nanmean(patients[:,2])
print(avg_sys_bp)
```

```python
# Find min , max , average Diastolic BP ignoring NaN
max_dia_bp = np.nanmax(patients[:,3])
print(max_dia_bp)

min_dia_bp = np.nanmin(patients[:,3])
print(min_dia_bp)

avg_dia_bp = np.nanmean(patients[:,3])
print(avg_dia_bp)
```

```python
# Find patients whose temperature is greater than 100°F
# d = np.where(np.isnan(patients),0,patients)

# print(d)
result = patients[patients[:,4] > 100]
print(result)
```

```python
# find critical pat. -> & |
# Temperature > 100
# OR
# Heart Rate > 90
# OR
# Systolic BP > 150
condition = (patients[:,4] > 100) | (patients[:,5] > 90) | (patients[:,2] > 150)
crit_pat = patients[condition]
print(crit_pat)
```
