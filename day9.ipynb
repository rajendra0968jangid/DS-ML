```python
import numpy as np

students = np.array([
    [101, 78, 85, 92, 88], # id,maths,english,computer,science
    [102, 65, 72, 70, 68],
    [103, 90, 95, 91, 94],
    [104, 45, 55, 60, 50],
    [105, 82, 79, 85, 80],
    [106, 55, 60, 58, 62],
    [107, 88, 92, 86, 90],
    [108, 72, 68, 75, 70],
    [109, 95, 91, 96, 94],
    [110, 40, 48, 45, 42]
])

print(students)
```

```python
# shape,size,dimension
print("shape:",students.shape) # 10 rows, 5 cols
print("size:",students.size) # no. of elements
print("Dimensions:",students.ndim) # 2 d
```

```python
# access single student
print(students[0]) # math , eng , com, sci
```

```python
# specific marks math , eng, com, sci
print(students[0,1])
```

```python
# specific marks math col and row 3
# specific marks eng col and row 5
# specific marks com col and row 6
print(students[2,1])
print(students[4,2])
print(students[5,3])
```

```python
# First 3 students
print(students[:3])
```

```python
# Last 3 students
print(students[-3:])
```

```python
# First 3 students with id and math
print(students[:3,0:2]) # row , cols
```

```python
# extract maths
maths = students[:,1] # all rows , 1 cols
print(maths)
```

```python
# First 5 Students marks
print(students[:5,1:])
```

```python
# maths , eng, com, sci
# maths and eng
print(students[:,1:3])
```

```python
# com and sci
print(students[:,3:5])
```

```python
# reverse dataset
print(students[::-1])
```

```python
# reverse subject order
print(students[:,1:][:,::-1])
```

```python
# total | sum of marks of all subject and particular student
marks = students[:,1:]
total = np.sum(marks,axis=1) # by default row but we need cols
print(total)

# alternative
for i in range(len(marks)):
  count = 0
  for j in range(4):
    count += marks[i][j]
  print(f"student {i} marks {count}")
```
