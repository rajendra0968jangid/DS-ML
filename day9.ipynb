```python
# np.where() | maths >= 50 -> pass another fail
marks = students[:,1]
result = np.where(marks >= 50,"Pass","Fail")
print(result)
```

```python
# np.where() | eng > 90 -> a grade | eng > 75 -> b grade | eng > 60 -> c grade | less then 60 fail
marks = students[:,2]
grade = np.where(marks >= 90 , "A",np.where(marks >= 75 , "B", np.where(marks >=60, "c","Fail")))

print(marks)
print(grade)

# (condition)?"true":(condition)?"true":(condition):"true":false
```

```python
# np.where() | 10 marks bonus | eng
marks = students[:,2]
new_marks = np.where(marks < 60, marks+10,marks)
print(new_marks)

grade = np.where(new_marks >= 90 , "A",np.where(new_marks >= 75 , "B", np.where(new_marks >=60, "c","Fail")))

print(new_marks)
print(grade)
```

```python
import numpy as np
students_nan = np.array([
    [101, 78, 85, 92, 88],
    [102, 65, np.nan, 70, 68],
    [103, 90, 95, 91, 94],
    [104, 45, 55, np.nan, 50],
    [105, 82, 79, 85, 80],
    [106, np.nan, 60, 58, 62],
    [107, 88, 92, 86, 90],
    [108, 72, 68, 75, 70],
    [109, 95, 91, 96, 94],
    [110, 40, 48, 45, 42]
])
print(students_nan)
```

```python
# find nan values -> isnan()
print(np.isnan(students_nan))
```

```python
# count of nan values
print(np.sum(np.isnan(students_nan)))
```

```python
# np.where() -> find position of nan values | row and cols
result = np.where(np.isnan(students_nan))
print(result)
#
print("rows",result[0])
print("cols",result[1])

# alternative
rows,cols = np.where(np.isnan(students_nan))
print("rows",rows)
print("cols",cols)
```

```python
# missing values subject-wise | math , eng, com, sci
marks = students_nan[:,1:]
missing = np.sum(np.isnan(marks),axis=0)
print(missing)
```

```python
# replace nan with 0
cleaned = students_nan.copy()
cleaned[np.isnan(cleaned)] = 0
print(cleaned)
```

```python
# replace nan with subject average -> np.nanmean()
cleaned = students_nan.copy()
marks = cleaned[:,1:]
for i in range(4):
  sub = marks[:,i]
  avg = np.nanmean(sub)
  sub[np.isnan(sub)] = np.ceil(avg)

print(cleaned)
```

```python
# maths marks ascending and decending  -> np.sort
import numpy as np
students_nan = np.array([
    [101, 78, 85, 92, 88],
    [102, 65, np.nan, 70, 68],
    [103, 90, 95, 91, 94],
    [104, 45, 55, np.nan, 50],
    [105, 82, 79, 85, 80],
    [106, np.nan, 60, 58, 62],
    [107, 88, 92, 86, 90],
    [108, 72, 68, 75, 70],
    [109, 95, 91, 96, 94],
    [110, 40, 48, 45, 42]
])
print(students_nan)
```

```python
maths = students_nan[:,1]
print(maths)

print(np.sort(maths)) # ascending
print(np.sort(maths)[::-1]) # decending
```

```python
# argsort | sort complete students by maths
# 101 , 95, 80,
# 102 , 80 , 85
# 103 , 98 , 90

# 102,80,85
# 101,95,
# 103,98,
index = np.argsort(students_nan[:,1])
print(index)
result = students_nan[index]
print(result)

# use argmin | argmax
print(np.argmin(students_nan[:,0]))
```

```python
# vstack | hstack | delete
import numpy as np

students = np.array([
    [101, 78, 85, 92, 88], # id,maths,english,computer,science
    [102, 65, 72, 70, 68],
    [103, 90, 95, 91, 94],
    [104, 45, 55, 60, 50],
    [105, 82, 79, 85, 80],
    [106, 55, 60, 58, 62],
    [107, 88, 92, 86, 90],
    [108, 72, 68, 75, 70],
    [109, 95, 91, 96, 94],
    [110, 40, 48, 45, 42]
])

print(students)
```

```python
# add new column of total -> hstack
marks = students[:,1:]
# print(marks)

total = np.sum(marks,axis=1)
# print(total)

total_column = total.reshape(-1,1)

# print(total_column)

result = np.hstack((marks,total_column))

print(result)
```

```python
# Add percentage column -> [::-1] | -> (-1,1)
marks = students[:,1:]
# print(marks)

total = np.sum(marks,axis=1)
# print(total)

# print(len(marks))
total_column = total.reshape(len(marks),1)

# print(total_column)

per = total/400*100

per_column = per.reshape(-1,1)

result = np.hstack((marks,total_column,per_column))

result = result.astype(int)
print(result)
print(result.dtype)  #
```
