```python
import numpy as np

students = np.array([
    [101, 78, 85, 92, 88], # id,maths,english,computer,science
    [102, 65, 72, 70, 68],
    [103, 90, 95, 91, 94],
    [104, 45, 55, 60, 50],
    [105, 82, 79, 85, 80],
    [106, 55, 60, 58, 62],
    [107, 88, 92, 86, 90],
    [108, 72, 68, 75, 70],
    [109, 95, 91, 96, 94],
    [110, 40, 48, 45, 42]
])

print(students)
```

```python
# shape,size,dimension
print("shape:",students.shape) # 10 rows, 5 cols
print("size:",students.size) # no. of elements
print("Dimensions:",students.ndim) # 2d
```

```python
# access single student
print(students[0]) # math , eng , com, sci
```

```python
# specific marks math , eng, com, sci
print(students[0,1]) # row =0 and cols =1
```

```python
# specific marks math col and row 3
# specific marks eng col and row 5
# specific marks com col and row 6
print(students[2,1])
print(students[4,2])
print(students[5,3])
```

```python
# First 3 students
print(students[:3])
```

```python
# Last 3 students
print(students[-3:])
```

```python
# First 3 students with id and math
print(students[:3,0:2]) # row , cols
```

```python
# extract maths
maths = students[:,1] # all rows , 1 cols
print(maths)
```

```python
# First 5 Students marks
print(students[:5,1:])
```

```python
# maths , eng, com, sci
# maths and eng
print(students[:,1:3])
```

```python
# com and sci
print(students[:,3:5])
```

```python
# reverse dataset
print(students[::-1])
```

```python
# reverse subject order
print(students[:,1:][:,::-1])
```

```python
# total | sum of marks of all subject and particular student
marks = students[:,1:]
total = np.sum(marks,axis=1) # by default row but we need cols
print(total)

# alternative
for i in range(len(marks)):
  count = 0
  for j in range(4):
    count += marks[i][j]
  print(f"student {i} marks {count}")
```

```python
# Average marks
marks = students[:,1:]  #

avg = np.mean(marks,axis=1) # by default row wise | for cols use axis=1
print(avg)
```

```python
# Calculate percentage ex. ([78 + 85 + 92 + 88]/400 * 100)
marks = students[:,1:]
per = np.sum(marks,axis=1)/400*100
print(per)
```

```python
# Highest and Lowest maths marks
marks = students[:,1]
min = np.min(marks)
max = np.max(marks)
print(f"minimum marks {min} and maximum marks {max}")
```

```python
# Average of each subject
marks = students[:,1:]
avg = np.mean(marks,axis=0)
print(avg)
```

```python
# maximum and minimum of each subjects
marks = students[:,1:]
min = np.max(marks,axis=0)
max = np.min(marks,axis=0)

print(min,max)
```

```python
# standard deviation of each subject
marks = students[:,1:]
dev = np.std(marks,axis=0)
print(dev)
```

```python
# np.where() | maths >= 50 -> pass another fail
marks = students[:,1]
result = np.where(marks >= 50,"Pass","Fail")
print(result)
```

```python
# np.where() | eng > 90 -> a grade | eng > 75 -> b grade | eng > 60 -> c grade | less then 60 fail
marks = students[:,2]
grade = np.where(marks >= 90 , "A",np.where(marks >= 75 , "B", np.where(marks >=60, "c","Fail")))

print(marks)
print(grade)

# (condition)?"true":(condition)?"true":(condition):"true":false
```

```python
# np.where() | 10 marks bonus | eng
marks = students[:,2]
new_marks = np.where(marks < 60, marks+10,marks)
print(new_marks)

grade = np.where(new_marks >= 90 , "A",np.where(new_marks >= 75 , "B", np.where(new_marks >=60, "c","Fail")))

print(new_marks)
print(grade)
```
