```python
import matplotlib.pyplot as plt # plt work as nick name | alias

# pyplot is a collection of functions
```

create chart

```python
# data
x = [1,2,3,4,5]
y = [10,20,35,40,50]

plt.figure(figsize=(4,3)) # size of graph

plt.plot(x,y) # x-cor and y-cor | line

plt.show() # graph print
```

customize chart

```python
plt.figure(figsize=(4,3)) # hor | ver

plt.plot(x, y, color='g', marker='v', linestyle=':',linewidth=3, markersize=12)

plt.title("sales graph") #
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.show()
```

Advanced - multi line chart & legend

```python
# data
x = [2010,2015,2020,2025,2030]
y1 = [10,20,35,45,50]
y2 = [12,22,38,48,60]

plt.figure(figsize=(4,4))

plt.plot(x,y1,label="textile")
plt.plot(x,y2,label="furniture")

plt.title("Sales report") # title top show
plt.xlabel("Years") # x-cor | years
plt.ylabel("Sales") # y-cor | sales

plt.legend() # legent define the lines

plt.show()
```

```python
# data
x = [2010,2015,2020,2025,2030]
y1 = [10,20,35,45,50]
y2 = [12,22,38,48,60]

plt.figure(figsize=(4,4))

plt.plot(x,y1,label="textile",color='y', marker='v', linestyle=':',linewidth=3, markersize=12)
plt.plot(x,y2,label="furniture",color='b', marker='v', linestyle=':',linewidth=3, markersize=12)

plt.title("Sales report") # title top show
plt.xlabel("Years") # x-cor | years
plt.ylabel("Sales") # y-cor | sales

plt.legend() # legent define the lines

plt.show()
```

Bar chart

```python
x = ['a','b','c','d','e']
y = [10,20,30,40,50]

plt.figure(figsize=(4,3))
plt.bar(x,y)
plt.title("report")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.show()
```

Customize bar chart

```python
x = ['a','b','c','d','e']
y = [10,20,30,40,50]

plt.figure(figsize=(4,3))

plt.bar(x,y,color=["orange","blue","yellow","green","black"],
        edgecolor="purple",linewidth=2)

plt.title("report")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.show()
```

multi -bar chart

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4])
y1 = [10,20,30,40]
y2 = [20,40,50,55]

plt.figure(figsize=(6,4))

w = 0.40

plt.bar(x - w/2,y1,label="girls", width=w) # chart 1
plt.bar(x + w/2,y2,label="boys",width=w) # chart 2

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = [2000,2005,2010,2015,2020]


y1 = [10,20,30,40,50] # 0.33
y2 = [15,25,35,45,55]
y3 = [20,40,60,66,40]

x_pos = np.arange(len(x))
print(x_pos)

w = 0.25

plt.bar(x_pos-w,y1,w,label="Math") # 1 - 0.33 => 0.67
plt.bar(x_pos,y2,w,label="Eng") # 1          => 1
plt.bar(x_pos+w,y3,w,label="Sci") # 1 + 0.33 => 1.33

plt.xticks(x_pos,x)
plt.legend()
plt.grid(True)
plt.show()
```

Histogram

```python
import matplotlib.pyplot as plt # hito -> cont.

data = [2,5,4,3,5,6,7,7,5,3,5,7,10,14,19] # rating
plt.figure(figsize=(4,4))
plt.hist(data,bins=3) # bad , average, good

plt.title("hist exam")
plt.xlabel("values")
plt.ylabel("freque")

plt.show()
```

pie chart

```python
categories = ['A','B','C','D','E']
sales = [10,20,30,40,50]

plt.pie(sales,labels=categories, autopct="%1.1f%%", startangle=0,hatch=['**O', 'oO', 'O.O', '.||.'])
plt.show()
```

subplot

```python
# data -1 -> bar chart
categories = ['Mon','Tue','Wed','Thu','Fri']
sales = [10,20,30,40,50]

# data 2 -> scatter plot
x = [2000,2010,2020,2030,2040]
y = [20,40,60,80,100]

plt.figure(figsize=(6,5))

# first plot -> bar chart
plt.subplot(1,2,1) # row , columns , 1
plt.bar(categories,sales)

# second plot -> scatter plot
plt.subplot(1,2,2) # row , columns , 2
plt.scatter(x,y)

plt.show()
```

```python
# data -1 -> bar chart
categories = ['Mon','Tue','Wed','Thu','Fri']
sales = [10,20,30,40,50]

# data 2 -> scatter plot
x = [2000,2010,2020,2030,2040]
y = [20,40,60,80,100]

# plt.figure(figsize=(6,5))

# first plot -> bar chart
plt.subplot(2,2,1) # row , columns , 1
plt.bar(categories,sales)
plt.title("chart box 1")

# second plot -> scatter plot
plt.subplot(2,2,2) # row , columns , 2
plt.scatter(x,y)
plt.title("chart box 2")

# first plot -> bar chart
plt.subplot(2,2,3) # row , columns , 3
plt.bar(categories,sales)
plt.title("chart box 3")

# second plot -> scatter plot
plt.subplot(2,2,4) # row , columns , 4
plt.scatter(x,y)
plt.title("chart box 4")

plt.tight_layout()
plt.savefig("subplot.png")
plt.show()
```
