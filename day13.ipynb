```python
import matplotlib.pyplot as plt # plt work as nick name | alias

# pyplot is a collection of functions
```

create chart

```python
# data
x = [1,2,3,4,5]
y = [10,20,35,40,50]

plt.figure(figsize=(4,3)) # size of graph

plt.plot(x,y) # x-cor and y-cor

plt.show() # graph print
```

customize chart

```python
plt.figure(figsize=(4,3)) # hor | ver

plt.plot(x, y, color='g', marker='v', linestyle=':',linewidth=3, markersize=12)

plt.title("sales graph") #
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.show()
```

Advanced - multi line chart & legend

```python
# data
x = [2010,2015,2020,2025,2030]
y1 = [10,20,35,45,50]
y2 = [12,22,38,48,60]

plt.figure(figsize=(4,4))

plt.plot(x,y1,label="textile")
plt.plot(x,y2,label="furniture")

plt.title("Sales report") # title top show
plt.xlabel("Years") # x-cor | years
plt.ylabel("Sales") # y-cor | sales

plt.legend() # legent define the lines

plt.show()
```
